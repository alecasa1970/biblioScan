## 3. Buscar Artigos por Palavra-Chave

```python
def buscar_artigos(termo: str) -> list:
    """Busca artigos por título, autores, abstract ou palavras-chave."""
    termo_lower = termo.lower()
    resultados = []
    for art in ARTICLES_DB:
        campos = [
            art['title'].lower(),
            art['authors'].lower(),
            art['abstract'].lower(),
            ' '.join(art['keywords']).lower()
        ]
        if any(termo_lower in campo for campo in campos):
            resultados.append(art)
    return resultados

# Exemplos de busca
for termo in ['alienação', 'Engels', 'capital', 'família']:
    res = buscar_artigos(termo)
    print(f"🔍 '{termo}': {len(res)} resultado(s)")
    for r in res:
        print(f"   → [{r['year']}] {r['title'][:70]}...")
    print()
```

## 4. Upload de Novo Artigo

Adicione um novo artigo à base de dados fornecendo os metadados abaixo.

```python
def upload_artigo(
    title: str,
    authors: str,
    year: int,
    journal: str,
    abstract: str,
    keywords: list,
    doi: str = "N/A",
    access: str = "Não informado",
    relevance: str = "Média"
) -> dict:
    """Faz o upload (inserção) de um novo artigo na base de dados."""
    novo_id = max(a['id'] for a in ARTICLES_DB) + 1
    novo_artigo = {
        "id": novo_id,
        "title": title,
        "authors": authors,
        "year": year,
        "journal": journal,
        "abstract": abstract,
        "keywords": keywords,
        "doi": doi,
        "access": access,
        "relevance": relevance,
        "uploaded_at": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }
    ARTICLES_DB.append(novo_artigo)
    
    # Salvar base atualizada em JSON
    with open('articles_db.json', 'w', encoding='utf-8') as f:
        json.dump(ARTICLES_DB, f, indent=2, ensure_ascii=False)
    
    print(f'✅ Artigo #{novo_id} adicionado com sucesso!')
    print(f'   Título: {title}')
    print(f'   Autores: {authors} ({year})')
    print(f'   Total na base: {len(ARTICLES_DB)} artigos')
    return novo_artigo


# ─── EXEMPLO DE USO ─────────────────────────────────────────────────────────
# Descomente e preencha para adicionar um novo artigo:

# novo = upload_artigo(
#     title="Seu Título Aqui",
#     authors="Nome do Autor",
#     year=2024,
#     journal="Nome do Periódico / Editora",
#     abstract="Resumo do artigo...",
#     keywords=["palavra1", "palavra2"],
#     doi="10.xxxx/xxxxx",
#     access="Acesso aberto",
#     relevance="Alta"
# )


# ─── DEMO: adicionando um 11º artigo ────────────────────────────────────────
upload_artigo(
    title="Karl Marx and the Digital Age: Relevance of Marxist Theory in the 21st Century",
    authors="Christian Fuchs",
    year=2020,
    journal="Routledge — Digital Labour and Karl Marx",
    abstract="Fuchs aplica a teoria marxista do trabalho e mais-valia ao contexto da economia digital, analisando como plataformas como Google e Facebook exploram o trabalho digital dos usuários. Introduz o conceito de 'trabalho digital' como nova forma de exploração capitalista.",
    keywords=["Marx", "trabalho digital", "capitalismo de plataforma", "mais-valia digital", "século XXI"],
    doi="10.4324/9781315880006",
    access="Routledge",
    relevance="Alta"
)
```

## 8. Exportar Resumo dos Artigos como HTML
